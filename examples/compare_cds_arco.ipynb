```python
import matplotlib.pyplot as plt
import xarray as xr
```

```python
ds_old = xr.open_dataset('ls2d_era5.nc')
ds_new = xr.open_dataset('ls2d_era5_arco.nc')
```

```python
# Plot variables as example:
nrow = 5
ncol = 2
sp = 1

t = 0

plt.figure(figsize=(8, 8))
plt.subplot(nrow, ncol, sp)
sp += 1
plt.plot(ds_old['thl'][t], ds_old['z'], label='CDS')
plt.plot(ds_new['thl'][t], ds_new['z'], label='ARCO')
plt.xlabel(r'$\theta_l$ (K)')
plt.legend()

plt.subplot(nrow, ncol, sp)
sp += 1
plt.plot(ds_old['dtthl_advec'][t] * 3600, ds_old['z'])
plt.plot(ds_new['dtthl_advec'][t] * 3600, ds_new['z'])
plt.xlabel(r'$\partial \theta_l/\partial t$ advec (K h$^{-1}$)')

plt.subplot(nrow, ncol, sp)
sp += 1
plt.plot(ds_old['qt'][t] * 1e3, ds_old['z'])
plt.plot(ds_new['qt'][t] * 1e3, ds_new['z'])
plt.xlabel(r'$q_t$ (g kg$^{-1}$)')

plt.subplot(nrow, ncol, sp)
sp += 1
plt.plot(ds_old['dtqt_advec'][t] * 3600000, ds_old['z'])
plt.plot(ds_new['dtqt_advec'][t] * 3600000, ds_new['z'])
plt.xlabel(r'$\partial q_t/\partial t$ advec (g kg$^{-1}$ h$^{-1}$)')

plt.subplot(nrow, ncol, sp)
sp += 1
plt.plot(ds_old['u'][t], ds_old['z'])
plt.plot(ds_new['u'][t], ds_new['z'])
plt.xlabel(r'$u$ (m s$^{-1}$)')

plt.subplot(nrow, ncol, sp)
sp += 1
plt.plot(ds_old['dtu_advec'][t] * 3600, ds_old['z'])
plt.plot(ds_new['dtu_advec'][t] * 3600, ds_new['z'])
plt.xlabel(r'$\partial u/\partial t$ advec (m s$^{-1}$ h$^{-1}$)')

plt.subplot(nrow, ncol, sp)
sp += 1
plt.plot(ds_old['v'][t], ds_old['z'])
plt.plot(ds_new['v'][t], ds_new['z'])
plt.xlabel(r'$v$ (m s$^{-1}$)')

plt.subplot(nrow, ncol, sp)
sp += 1
plt.plot(ds_old['dtv_advec'][t] * 3600, ds_old['z'])
plt.plot(ds_new['dtv_advec'][t] * 3600, ds_new['z'])
plt.xlabel(r'$\partial v/\partial t$ advec (m s$^{-1}$ h$^{-1}$)')

plt.subplot(nrow, ncol, sp)
sp += 1
plt.plot(ds_old['ug'][t], ds_old['z'])
plt.plot(ds_new['ug'][t], ds_new['z'])
plt.xlabel(r'$u_g$ (m s$^{-1}$)')

plt.subplot(nrow, ncol, sp)
sp += 1
plt.plot(ds_old['vg'][t], ds_old['z'])
plt.plot(ds_new['vg'][t], ds_new['z'])
plt.xlabel(r'$v_g$ (m s$^{-1}$)')

plt.tight_layout()
```

```python
# Time series surface:
nrow = 2
ncol = 2
sp = 1

plt.figure(figsize=(8, 5))
plt.subplot(nrow, ncol, sp)
sp += 1
plt.plot(ds_old['time'], ds_old['wth'], label='CDS')
plt.plot(ds_new['time'], ds_new['wth'], label='ARCO')
plt.ylabel(r'$w\theta_s$ (K m s$^{-1}$)')
plt.legend()

plt.subplot(nrow, ncol, sp)
sp += 1
plt.plot(ds_old['time'], ds_old['wq'] * 1e3)
plt.plot(ds_new['time'], ds_new['wq'] * 1e3)
plt.ylabel(r'$wq_s$ (g kg$^{-1}$ m s$^{-1}$)')

plt.subplot(nrow, ncol, sp)
sp += 1
plt.plot(ds_old['time'], ds_old['ts'])
plt.plot(ds_new['time'], ds_new['ts'])
plt.ylabel(r'$T_s$ (K)')

plt.subplot(nrow, ncol, sp)
sp += 1
plt.plot(ds_old['time'], ds_old['ps'] / 100)
plt.plot(ds_new['time'], ds_new['ps'] / 100)
plt.ylabel(r'$p_s$ (hPa)')

plt.tight_layout()
```
